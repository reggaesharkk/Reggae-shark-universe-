```python
import json
from pathlib import Path
rows=json.loads(Path("/content/wp11_scaling.json").read_text())
for r in rows:
    print(f"N={r['base_cutoff']} to N={r['scaled_cutoff']} | K={r['base_threshold']} to K={r['scaled_threshold']}")
    print("time:",r["base_time_end"],"to",r["scaled_time_end"])
    print("integral:",r["base_signed_integral"],"to",r["scaled_signed_integral"])
    print("predicted scaled integral:",r["predicted_integral"])
    print("largest normalized check error:",max(r["maximum_relative_errors"].values()))
    print()
```
